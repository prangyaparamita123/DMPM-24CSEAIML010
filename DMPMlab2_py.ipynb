```python
def median(data):
    data.sort()
    n = len(data)
    if n % 2 == 1:
        return data[n // 2]
    else:
        return (data[n // 2 - 1] + data[n // 2]) / 2
data = list(map(int, input("Enter data points separated by spaces: ").split()))
result = median(data)
print("Median =", result)
```

```python
n = int(input("Enter the number of data points: "))
data = []
for i in range(n):
    num = int(input("Enter data point: "))
    data.append(num)
mode = data[0]
max_count = 0
for i in data:
    count = 0
    for j in data:
        if i == j:
            count = count + 1
    if count > max_count:
        max_count = count
        mode = i
print("Mode =", mode)
```

```python
def mode(data):
    max_count = 0
    mode_value = data[0]
    for i in data:
        count = 0
        for j in data:
            if i == j:
                count += 1
        if count > max_count:
            max_count = count
            mode_value = i

    return mode_value
data = list(map(int, input("Enter data points separated by spaces: ").split()))
result = mode(data)
print("Mode =", result)
```

```python
n = int(input("Enter the number of data points: "))
data = []
total = 0
for i in range(n):
    num = int(input("Enter data point: "))
    data.append(num)
    total = total + num
mean = total / n
sum_sq = 0
for i in data:
    sum_sq = sum_sq + (i - mean) ** 2
variance = sum_sq / n
print("Variance =", variance)
```

```python
def variance(data):
    n = len(data)
    mean = sum(data) / n
    total = 0
    for i in data:
        total = total + (i - mean) ** 2
    return total / n
data = list(map(int, input("Enter data points separated by spaces: ").split()))
result = variance(data)
print("Variance =", result)
```
